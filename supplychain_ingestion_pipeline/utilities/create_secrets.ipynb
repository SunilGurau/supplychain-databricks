```python
from databricks.sdk import WorkspaceClient

w = WorkspaceClient()
```

```python
# Check if scope exists
try:
    w.secrets.create_scope(scope="kafka")
except:
    print("Scope 'kafka' already exists")
```

```python
w.secrets.put_secret(
    scope="kafka",
    key="keystore-password",
    string_value="PASSWORD_HERE"
)
print("Secret keystore-password created successfully")
```

```python
w.secrets.put_secret(
    scope="kafka",
    key="truststore-password",
    string_value="PASSWORD_HERE"
)
print("Secret truststore-password created successfully")
```

```python
w.secrets.put_secret(scope="kafka", key="key-password", string_value="PASSWORD_HERE")
print("Secret key-password created successfully")
```
