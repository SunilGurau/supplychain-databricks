```python
%sh
# Create truststore from CA certificate
keytool -import -trustcacerts -alias ca-cert \
  -file /Volumes/operations-catalog/supplychain/kafka-client-certificates/ca.pem \
  -keystore /Volumes/operations-catalog/supplychain/kafka-client-certificates/truststore.jks \
  -storepass $TRUSTSTORE_PASSWORD\
  -noprompt
```

```python
%sh
# Create keystore from client certificate and key
# First, convert to PKCS12 format
openssl pkcs12 -export \
  -in /Volumes/operations-catalog/supplychain/kafka-client-certificates/service.cert \
  -inkey /Volumes/operations-catalog/supplychain/kafka-client-certificates/service.key \
  -out /Volumes/operations-catalog/supplychain/kafka-client-certificates/keystore.p12 \
  -name kafka-client \
  -passout pass:$KEYSTORE_PASSWORD
```

```python
%sh
# Then convert PKCS12 to JKS
keytool -importkeystore \
  -srckeystore /Volumes/operations-catalog/supplychain/kafka-client-certificates/keystore.p12 \
  -srcstoretype PKCS12 \
  -srcstorepass $KEYSTORE_PASSWORD \
  -destkeystore /Volumes/operations-catalog/supplychain/kafka-client-certificates/keystore.jks \
  -deststoretype JKS \
  -deststorepass $KEYSTORE_PASSWORD \
  -destkeypass $KEY_PASSWORD
```
